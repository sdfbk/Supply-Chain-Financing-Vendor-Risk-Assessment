```python
def flexible_date_cleaner(date_str):
    if pd.isna(date_str) or str(date_str).strip() == "":
        return pd.NaT

    # Try 1: Explicitly try Day-First (e.g., 25/12/2025)
    # This silences the warning because we are telling Pandas what to expect.
    try:
        return pd.to_datetime(date_str, dayfirst=True, errors='raise')
    except:
        # Try 2: Try Year-First or Month-First (e.g., 2025-12-25)
        try:
            return pd.to_datetime(date_str, dayfirst=False, errors='raise')
        except:
            # Try 3: Mixed/Text format (e.g., "Jan 5, 2025")
            try:
                return pd.to_datetime(date_str, format='mixed', errors='raise')
            except:
                return pd.NaT

# Apply the fixed function
df_erp['Order_Date'] = df_erp['Order_Date'].apply(flexible_date_cleaner)
```

```python
risk_name = df_risk["Entity_Name"].to_list()
```

```python
def match_vendor(name):
  match,score = process.extractOne(str(name),risk_name)
  return match if score > 70 else "Unknown"
```

```python
df_erp["Vendor_Name"] = df_erp["Vendor_Name"].apply(match_vendor)
```
